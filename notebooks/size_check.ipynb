```python
import sys

num = int(1)

char1 = "A"
char2 = "AB"
char3 = "ABC"

num_size = sys.getsizeof( num )
char1_size = sys.getsizeof( char1 )
char2_size = sys.getsizeof( char2 )
char3_size = sys.getsizeof( char3 )

print( f"Integer: {num_size}")
print( f"1 Char: {char1_size}")
print( f"2 Char: {char2_size}")
print( f"3 Char: {char3_size}")

print( f"Empty Char: {sys.getsizeof( '' )}" )
```

```python
print( f"Code:    {sys.getsizeof( 138 )}")
print( f"Airport: {sys.getsizeof( 'LaGuardia Airport' )}")
print( f"None:    {sys.getsizeof( 'None' )}")
```
