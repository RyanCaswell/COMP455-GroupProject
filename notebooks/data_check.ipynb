```python
import pandas as pd

df = pd.read_parquet( "../data/yellow_tripdata_2026-01.parquet" )

columns_to_check = [
    "tpep_pickup_datetime",
    "PULocationID",
    "DOLocationID",
    "tip_amount",
    "total_amount",
]

len( df )
```

```python
# Print all possible columns
for column in df.columns:
    print( column )
```

```python
invalid_dates = (
    (df["tpep_pickup_datetime"] < "2025-12-30") |
    (df["tpep_pickup_datetime"] >= "2026-02-02") |
    (df["tpep_pickup_datetime"].isna())
)

print(f"Dates outside January 2026: {invalid_dates.sum()}")
```

```python
invalid_pu = (
    (df["PULocationID"] < 1) |
    (df["PULocationID"] > 265) |
    (df["PULocationID"].isnull())
)

invalid_do = (
    (df["DOLocationID"] < 1) |
    (df["DOLocationID"] > 265) |
    (df["DOLocationID"].isnull())
)

print(f"Invalid pickup LocationIDs: {invalid_pu.sum()}")
print(f"Invalid drop-off LocationIDs: {invalid_do.sum()}")
```

```python
invalid_tips = (
    (df["tip_amount"] < 0) |
    (df["tip_amount"].isnull())
)

invalid_total = (
    (df["total_amount"] <= 0) |
    (df["total_amount"].isnull())
)

print(f"Invalid tip amounts: {invalid_tips.sum()}")
print(f"Invalid total amounts: {invalid_total.sum()}")
```
